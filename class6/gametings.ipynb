```python
# Okay so now I want to make an actual character.
# I'm keeping the stats pretty basic for now because
# I don't want this to become too complicated.

player = {
    "name": "Mars Explorer",
    "health": 100,
    "oxygen": 100,
    "crystals": 0,
    "attack": 15,
    "backpack": []
}

print("Welcome,", player["name"])
print("Health:", player["health"])
print("Oxygen:", player["oxygen"])
print("Backpack:", player["backpack"])
```

```python
# Now for the fun part! I want players to discover
# random items while exploring Mars.

# Some items will be useful and some will just be
# weird artifacts that players can collect.

items = [
    "Oxygen Tank",
    "Medkit",
    "Ancient Martian Coin",
    "Glowing Crystal",
    "Alien Compass",
    "Rusty Space Key",
    "Mysterious Fossil",
    "Plasma Sword"
]

# Let's randomly discover something!
found_item = random.choice(items)

print("You discovered:", found_item)

# Adding the item to my backpack
player["backpack"].append(found_item)

print("Your backpack:", player["backpack"])
```

```python
# I want the player to be able to check their backpack
# whenever they want, kind of like opening an inventory
# in a video game.

def open_backpack():
    print("\nYOUR BACKPACK")

    if len(player["backpack"]) == 0:
        print("Your backpack is empty!")
    else:
        for item in player["backpack"]:
            print("-", item)

open_backpack()
```

```python
# Collecting items is cute but I want them to
# actually DO something.

def use_oxygen_tank():
    if "Oxygen Tank" in player["backpack"]:
        player["oxygen"] = min(100, player["oxygen"] + 30)

        player["backpack"].remove("Oxygen Tank")

        print("You used an Oxygen Tank!")
        print("Oxygen:", player["oxygen"])
    else:
        print("You don't have an Oxygen Tank!")

# Test with this when you have an Oxygen Tank:
# use_oxygen_tank()
```
