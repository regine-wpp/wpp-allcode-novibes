```python
# Using enumerate
names = ["Alice", "Bob"]
for index, name in enumerate(names):
    print(f"Index {index}: {name}")

# Using dictionary items
user = {"username": "jdoe", "role": "admin"}
for key, value in user.items():
    print(f"{key} is {value}")

for number in range(1, 6):
    if number == 3:
        continue  # Skip 3
    if number == 5:
        break     # Stop the loop at 5
    print(number)
#break: Exits the loop entirely before it has finished iterating.
#continue: Skips the current iteration and jumps straight to the next one.
#pass: Acts as a null placeholder to avoid errors if you have an empty loop body.


```

```python
#List Comprehension
# The syntax template:
# new_list = [expression for item in iterable if condition]
#traditional for loop: 
#Traditional for Loop	List Comprehension
#squares = []
#for x in range(1, 4):
    #squares.append(x**2)	squares = [x**2 for x in range(1, 4)]

#Output: [1,4,9] For both

#Adding an if Statement (Filtering) You can append an if statement to the very end of the comprehension to filter out specific items.

numbers = [1, 2, 3, 4, 5, 6]

# Get only the even numbers
evens = [x for x in numbers if x % 2 == 0]
print(evens)  # Output: [2, 4, 6]


```
